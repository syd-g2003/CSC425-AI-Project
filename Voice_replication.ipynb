```python
#Import libraries
import librosa
from IPython.display import display, Audio
import torch
from TTS.api import TTS
from pathlib import Path
```

```python
#Ask the user to enter the file name for the voice they wish to replicate
while True:
    try:
        filename= input('Enter the filename of the audio')
        #Load and play audio file
        audio, sr=librosa.load(f'audio/Original/{filename}', sr=None)
        display(Audio(audio,rate=sr))
        print('Success!')
        break
    except FileNotFoundError:
        print('You entered a wrong filename. Try again!')
    except Exception as e:
        print("An error occured. Try again!")
```

```python
#Initialize the xtts-v2 model
model_name='tts_models/multilingual/multi-dataset/xtts_v2'
model=TTS(model_name).to('cpu')
```

```python
output_filename= Path(filename).stem + '_replicated.wav'
processed_filename= Path(filename).stem + '_processed.wav'
processed_output_filename=Path(processed_filename).stem + '_replicated.wav'
```

```python
#Generate an output audio files that read a text in the same voice as the input file

model.tts_to_file(
    text='Although the weather was unpredictable, we decided to continue our journey. Sometimes, the most interesting experiences happen when things do not go exactly as planned',
    file_path=f'audio/Replicated/{output_filename}',
    speaker_wav=f'audio/Original/{filename}',
    language='en'
)
```

```python
#Generate an output audio files that read a text in the same voice as the processed input file
try:
    model.tts_to_file(
        text='Although the weather was unpredictable, we decided to continue our journey. Sometimes, the most interesting experiences happen when things do not go exactly as planned',
        file_path=f'audio/Replicated/{processed_output_filename}',
        speaker_wav=f'audio/Processed/{processed_filename}',
        language='en'
    )
except Exception as e:
    print("An error occured. Make sure to run the Preprocessing.ipynb on this file beforehand!")
```

```python
#Read generated audio file from original audio
output1,sr1=librosa.load(f'audio/Replicated/{output_filename}', sr=None)
display(Audio(output1,rate=sr1))

#Read generated audio file from preprocessed audio
output2,sr2=librosa.load(f'audio/Replicated/{processed_output_filename}', sr=None)
display(Audio(output2,rate=sr2))
```
